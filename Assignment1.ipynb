```python
import numpy as np

np.random.seed(60)

X = np.random.rand(10, 2)

A = X[:, np.newaxis, :]
B = X[np.newaxis, :, :]

D = A - B
D2 = D ** 2
distance = np.sum(D2, axis=2)

print("Coordinates shape:", A.shape)
print("Squared differences shape:", D2.shape)
print("Distance matrix shape:", distance.shape)

print("\nSquared Distance Matrix:")
print(distance)
```

```python
import numpy as np

np.random.seed(60)

X = np.random.rand(10, 2)

D = X[:, np.newaxis, :] - X[np.newaxis, :, :]
D2 = np.sum(D ** 2, axis=2)

print("Diagonal:")
print(np.diag(D2))

print("All diagonal values are zero:")
print(np.allclose(np.diag(D2), 0))
```

```python
import numpy as np

np.random.seed(60)

X = np.random.rand(10, 2)

D = X[:, np.newaxis, :] - X[np.newaxis, :, :]
D2 = np.sum(D ** 2, axis=2)

neighbors = np.argsort(D2, axis=1)

print("Neighbor ordering:")
print(neighbors)

print("\nFirst column:")
print(neighbors[:, 0])
```

```python
import numpy as np

np.random.seed(60)

X = np.random.rand(10, 2)

D = X[:, np.newaxis, :] - X[np.newaxis, :, :]
D2 = np.sum(D ** 2, axis=2)

neighbors = np.argsort(D2, axis=1)

K = 2

nearest = neighbors[:, 1:K+1]

print("Two nearest neighbors:")
print(nearest)
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(60)

X = np.random.rand(10, 2)

D = X[:, np.newaxis, :] - X[np.newaxis, :, :]
D2 = np.sum(D ** 2, axis=2)

neighbors = np.argsort(D2, axis=1)

plt.scatter(X[:, 0], X[:, 1])

for i in range(10):
    for j in neighbors[i, 1:3]:
        plt.plot([X[i, 0], X[j, 0]],
                 [X[i, 1], X[j, 1]])

plt.xlabel("X")
plt.ylabel("Y")
plt.title("2-Nearest Neighbor Network")
plt.show()
```

```python
import numpy as np

np.random.seed(60)

X = np.random.rand(10, 2)

K = 2

for i in range(10):
    distance = np.sum((X - X[i]) ** 2, axis=1)
    neighbors = np.argsort(distance)

    print("Point", i, ":", neighbors[1:K+1])
```

```python
import numpy as np
import time

for N in [100, 1000, 5000]:

    X = np.random.rand(N, 2)

    start = time.time()

    D = X[:, np.newaxis, :] - X[np.newaxis, :, :]
    D2 = np.sum(D ** 2, axis=2)
    neighbors = np.argsort(D2, axis=1)

    end = time.time()

    print("N =", N)
    print("Time =", end - start)
    print()
```

```python
import numpy as np
from scipy.spatial import KDTree

np.random.seed(60)

N = 10
D = 2
K = 2

X = np.random.rand(N, D)

tree = KDTree(X)

distance, neighbors = tree.query(X, k=K+1)

nearest = neighbors[:, 1:]

print("Two nearest neighbors:")
print(nearest)
```

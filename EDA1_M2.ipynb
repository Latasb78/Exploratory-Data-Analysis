```python
import pandas as pd

area = pd.Series([308252, 196024, 191791, 130060, 2109],
                 index=['Rajasthan', 'Madhya Pradesh', 'Maharashtra', 'Tamil Nadu', 'Goa'])

population = pd.Series([81, 72, 126, 77, 1.6],
                       index=['Rajasthan', 'Madhya Pradesh', 'Maharashtra', 'Tamil Nadu', 'Goa'])

states = pd.DataFrame({'area': area, 'population': population})
states['density'] = states['population'] / states['area']

print(states)
```

```python
import pandas as pd

indA = pd.Index([2, 3, 5, 7, 11])
indB = pd.Index([1, 3, 5, 7, 9])

print("Intersection:", indA.intersection(indB))
print("Union:", indA.union(indB))
print("Symmetric Difference:", indA.symmetric_difference(indB))
```

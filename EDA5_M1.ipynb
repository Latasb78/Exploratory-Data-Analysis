```python
import numpy as np

dtype = [('title', 'U30'), ('author', 'U20'), ('price', 'f4')]

books = np.array([
    ("Python Basics", "John Smith", 450.0),
    ("NumPy Guide", "David Lee", 350.0),
    ("Data Science", "Anita Rao", 550.0),
    ("Machine Learning", "Rahul Kumar", 600.0)
], dtype=dtype)

print(books)
```

```python
print(books['title'])
print(books[1])

amount = 500
print(books[books['price'] < amount]['title'])
```

```python
import numpy as np

dtype1 = np.dtype({
    'names': ['title', 'author', 'price'],
    'formats': ['U30', 'U20', 'f4']
})

dtype2 = np.dtype([
    ('title', 'U30'),
    ('author', 'U20'),
    ('price', 'f4')
])

print(dtype1)
print(dtype2)
print(dtype1 == dtype2)
```

```python
data_rec = books.view(np.recarray)

print(data_rec.price)
```

```python
import numpy as np

dtype = [('roll_no', 'i4'), ('name', 'U20'), ('marks', 'f4')]

students = np.array([
    (1, "Asha", 85.5),
    (2, "Rahul", 72.0),
    (3, "Sneha", 91.5),
    (4, "Kiran", 68.0),
    (5, "Anu", 88.0)
], dtype=dtype)

sorted_students = np.sort(students, order='marks')[::-1]

print(sorted_students)
```

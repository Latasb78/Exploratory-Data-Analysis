```python
import numpy as np

x = np.random.randint(1, 50, 10)
print(x)

y = np.sort(x)
print(y)
print(x)

x.sort()
print(x)
```

```python
i = np.argsort(x)
print(i)
print(x[i])
print(np.sort(x))
```

```python
import numpy as np

x = np.random.randint(1, 50, (4, 6))

print(x)
print(np.sort(x, axis=1))
print(np.sort(x, axis=0))
```

```python
import numpy as np

x = np.random.randint(1, 100, 15)

p = np.partition(x, 4)
print(p[:4])

print(np.sort(x)[:4])
```

```python
import numpy as np

student_names = np.array(["Asha", "Rahul", "Sneha", "Kiran", "Anu"])
scores = np.array([78, 65, 92, 55, 88])

i = np.argsort(scores)

print(student_names[i])
print(scores[i])
```

```python
import numpy as np

np.random.seed(0)

X = np.random.rand(10, 2)

dist_sq = np.sum((X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2, axis=-1)

i = np.argsort(dist_sq, axis=1)

print(i[:, 1:3])
```

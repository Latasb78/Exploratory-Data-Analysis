```python
import numpy as np

arr = np.arange(1, 13).reshape(3, 4)

print(arr[1, 2])
print(arr[-1])
print(arr[:, -1])
```

```python
import numpy as np

arr = np.arange(1, 13).reshape(3, 4)

print(arr[:2, :2])
print(arr[:, ::2])
print(arr[:, ::-1])
```

```python
import numpy as np

arr = np.arange(1, 13).reshape(3, 4)

view = arr[:3, :3]
view[0, 0] = 100

print(arr)
```

```python
import numpy as np

arr = np.arange(1, 13)

a = arr.reshape((12, 1))
b = arr[:, np.newaxis]

print(a)
print(b)
print(np.array_equal(a, b))
```

```python
import numpy as np

np.random.seed(0)

a = np.random.randint(0, 10, (2, 3))
b = np.random.randint(0, 10, (2, 3))

v1 = np.vstack((a, b))
v2 = np.concatenate((a, b), axis=0)

h1 = np.hstack((a, b))
h2 = np.concatenate((a, b), axis=1)

print(v1)
print(v2)
print(np.array_equal(v1, v2))

print(h1)
print(h2)
print(np.array_equal(h1, h2))
```

```python
import numpy as np

arr = np.arange(1, 16)

parts = np.split(arr, 3)

for part in parts:
    print(part)
```

```python
import numpy as np

arr = np.arange(1, 25).reshape(6, 4)

a = np.hsplit(arr, 2)
b = np.vsplit(arr, 2)

print(a[0])
print(a[1])

print(b[0])
print(b[1])
```
